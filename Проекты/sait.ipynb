# Информационная веб-система для танцевальных студий
## Full-stack проект
### Функциональность

1. Регистрация и авторизация пользователей;
2. Три типа пользователей: администратор, менеджер студии и клиент;
3. Управление студиями, залами, преподавателями и расписанием;
4. Бронирование танцевальных залов; запись на танцевальные занятия;
5. Проверка доступности времени; поиск занятий по дню недели, преподавателю и стилю танца; 
6. Обработка пользовательских действий через PHP; взаимодействие веб-приложения с MySQL.

### База данных:
1. Спроектировала структуру БД;
2. Создала таблицы пользователей, студий, преподавателей, залов, занятий, расписания и бронирований;
3. Реализовала связи между сущностями;
4. Использовала первичные и внешние ключи;
5. Написала SQL-запросы для создания и работы с данными.

### Мой функционал в проекте:
1. Разработка структуры и функциональности веб-приложения;
2. Реализация серверной логики;
3. Разработка базы данных;
4. Реализация фильтрации и поиска;
5. Реализация авторизации и разграничения ролей;
6. Разработка механизма бронирования;
7. Тестирование пользовательских сценариев;
8. Моделирование информационной системы.

# Используемые технологии: PHP, MySQL, SQL, HTML, UML/IDEF0.
