# Нейросетевой модуль обнаружения разведывательных сетевых атак
## Проект в рамках НИОКР с ООО «Газинформсервис», выпускная квалификационная работа
### Разработала информационно-аналитическое решение для обнаружения сетевой разведки.

1. Провела анализ предметной области и существующих подходов;
2. Разработала функциональную и информационную модели;
3. Выполнила обработку данных;
4. Обучила и сравнила архитектуры 1D CNN, 2D CNN, LSTM и CNN+LSTM;
5. Выбрана наиболее эффективная архитектура;
6. Разработан имитационный модуль с графическим интерфейсом;
7. Проведена апробация решения (регистрация патента на программу ЭВМ).



### Используемые технологии: Python, Machine Learning, Deep Learning, анализ данных.
