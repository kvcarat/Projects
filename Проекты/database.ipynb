# Проектирование базы данных и SQL
## Учебный проект по разработке информационной системы для бюро трудоустройства

1. анализ предметной области;
2. разработка инфологической и реляционной моделей;
3. проектирование структуры БД;
4. создание таблиц, ключей и связей;
5. написание SQL-запросов;
6. использование JOIN, GROUP BY, HAVING, подзапросов и UNION;
7. разработка пользовательских форм.

### Используемые технологии: SQL, MS Access, ER-моделирование.
